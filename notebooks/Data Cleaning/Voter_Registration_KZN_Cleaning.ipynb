# Data Cleaning: Voter Registration KZN

This notebook cleans and validates the KwaZulu-Natal voter registration dataset.

The raw files are kept unchanged in `data/raw/`. The cleaned files are saved separately in `data/interim/`.

The cleaning process checks for:
- Missing values
- Incorrect data types and number formatting
- Duplicate ward records
- Duplicate complete records
- Invalid ward codes
- Invalid voting district counts
- Invalid registered voter counts
- Preservation of records after cleaning
- Consistency between the raw and cleaned files

```python
import pandas as pd
from pathlib import Path
```

```python
# Notebook location:
# SPU-TEAM-DIRISA/notebooks/Data Collection/

project_root = Path("../..")

raw_folder = (
    project_root
    / "data"
    / "raw"
    / "Voter Registration_KZN"
)

interim_folder = (
    project_root
    / "data"
    / "interim"
    / "Voter Registration_KZN_cleaned"
)

interim_folder.mkdir(parents=True, exist_ok=True)

print("Raw folder:", raw_folder)
print("Raw folder exists:", raw_folder.exists())
print("Interim folder:", interim_folder)
```

```python
raw_files = sorted(raw_folder.glob("*.csv"))

print("Number of raw CSV files:", len(raw_files))

for file in raw_files:
    print(file.name)
```

```python
file_structure = []

for file in raw_files:
    df = pd.read_csv(file)

    file_structure.append({
        "file": file.name,
        "rows": len(df),
        "columns": len(df.columns),
        "column_names": list(df.columns)
    })

structure_df = pd.DataFrame(file_structure)

structure_df
```

```python
expected_columns = [
    "ward",
    "voting_districts",
    "registered_voters"
]

for file in raw_files:
    df = pd.read_csv(file)

    if list(df.columns) != expected_columns:
        print("Different structure:", file.name)

print("Structure check complete.")
```

```python
first_file = raw_files[0]

first_data = pd.read_csv(first_file)

print("File:", first_file.name)
print("Rows:", len(first_data))
print("Columns:", list(first_data.columns))

display(first_data.head())
```

```python
missing_results = []

for file in raw_files:
    df = pd.read_csv(file)

    for column in df.columns:
        missing = df[column].isnull().sum()

        missing_results.append({
            "file": file.name,
            "column": column,
            "missing_values": missing
        })

missing_df = pd.DataFrame(missing_results)

display(missing_df)

print(
    "Total missing values:",
    missing_df["missing_values"].sum()
)
```

```python
blank_results = []

for file in raw_files:
    df = pd.read_csv(file)

    for column in df.columns:
        blanks = (
            df[column]
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )

        blank_results.append({
            "file": file.name,
            "column": column,
            "blank_values": blanks
        })

blank_df = pd.DataFrame(blank_results)

display(blank_df)

print(
    "Total blank values:",
    blank_df["blank_values"].sum()
)
```

```python
ward_results = []

for file in raw_files:
    df = pd.read_csv(file)

    invalid_wards = (
        ~df["ward"]
        .astype(str)
        .str.strip()
        .str.match(r"^\d+$")
    ).sum()

    ward_results.append({
        "file": file.name,
        "invalid_ward_codes": invalid_wards
    })

ward_df = pd.DataFrame(ward_results)

display(ward_df)

print(
    "Total invalid ward codes:",
    ward_df["invalid_ward_codes"].sum()
)
```

```python
district_results = []

for file in raw_files:
    df = pd.read_csv(file)

    invalid_districts = (
        pd.to_numeric(
            df["voting_districts"],
            errors="coerce"
        )
        .le(0)
    ).sum()

    district_results.append({
        "file": file.name,
        "invalid_voting_districts": invalid_districts
    })

district_df = pd.DataFrame(district_results)

display(district_df)

print(
    "Total invalid voting district values:",
    district_df["invalid_voting_districts"].sum()
)
```

```python
duplicate_ward_results = []

for file in raw_files:
    df = pd.read_csv(file)

    duplicates = df["ward"].duplicated().sum()

    duplicate_ward_results.append({
        "file": file.name,
        "duplicate_wards": duplicates
    })

duplicate_ward_df = pd.DataFrame(duplicate_ward_results)

display(duplicate_ward_df)

print(
    "Total duplicate ward records:",
    duplicate_ward_df["duplicate_wards"].sum()
)
```

```python
# Check the data format and data types in all files

format_check = []

for file in raw_files:
    df = pd.read_csv(file)

    for column in df.columns:
        format_check.append({
            "file": file.name,
            "column": column,
            "data_type": str(df[column].dtype),
            "example_value": df[column].iloc[0]
        })

format_df = pd.DataFrame(format_check)

display(format_df)
```

```python
# Count registered voter values containing commas


# Inspect values in columns that are expected to contain numbers

for file in raw_files[:3]:
    df = pd.read_csv(file)

    print("\nFile:", file.name)

    print("Voting districts examples:")
    print(df["voting_districts"].head().to_list())

    print("Registered voters examples:")
    print(df["registered_voters"].head().to_list())
```

```python
# Fix the registered_voters format

cleaning_summary = []

for file in raw_files:

    df = pd.read_csv(file)

    # Remove commas and convert registered_voters to a numeric type
    df["registered_voters"] = (
        df["registered_voters"]
        .astype(str)
        .str.strip()
        .str.replace(",", "", regex=False)
        .astype("int64")
    )

    # Convert voting_districts to integer
    df["voting_districts"] = (
        pd.to_numeric(
            df["voting_districts"],
            errors="raise"
        ).astype(int)
    )

    # Keep ward as a string identifier
    df["ward"] = (
        df["ward"]
        .astype(str)
        .str.strip()
    )

    output_file = interim_folder / file.name

    df.to_csv(
        output_file,
        index=False,
        encoding="utf-8"
    )

    cleaning_summary.append({
        "file": file.name,
        "rows": len(df)
    })

print(f"Cleaning complete. {len(cleaning_summary)} files processed.")
```

```python
# Check the data types after cleaning

cleaned_format_check = []

for file in cleaned_files if "cleaned_files" in globals() else sorted(interim_folder.glob("*.csv")):

    df = pd.read_csv(file)

    for column in df.columns:
        cleaned_format_check.append({
            "file": file.name,
            "column": column,
            "data_type": str(df[column].dtype),
            "example_value": df[column].iloc[0]
        })

cleaned_format_df = pd.DataFrame(cleaned_format_check)

display(cleaned_format_df)
```

```python
duplicate_record_results = []

for file in raw_files:
    df = pd.read_csv(file)

    duplicates = df.duplicated().sum()

    duplicate_record_results.append({
        "file": file.name,
        "duplicate_complete_records": duplicates
    })

duplicate_record_df = pd.DataFrame(duplicate_record_results)

display(duplicate_record_df)

print(
    "Total duplicate complete records:",
    duplicate_record_df["duplicate_complete_records"].sum()
)
```

```python
row_counts = []

for file in raw_files:
    df = pd.read_csv(file)

    row_counts.append({
        "file": file.name,
        "rows": len(df)
    })

row_counts_df = pd.DataFrame(row_counts)

display(row_counts_df)

print(
    "Total ward records across all files:",
    row_counts_df["rows"].sum()
)
```

```python
cleaning_summary = []

for file in raw_files:

    df = pd.read_csv(file)

    cleaned_df = pd.DataFrame({
        "ward": df["ward"].astype(str).str.strip(),
        "voting_districts": (
            pd.to_numeric(
                df["voting_districts"],
                errors="raise"
            ).astype(int)
        ),
        "registered_voters": (
            df["registered_voters"]
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
            .astype("int64")
        )
    })

    output_file = interim_folder / file.name

    cleaned_df.to_csv(
        output_file,
        index=False,
        encoding="utf-8"
    )

    cleaning_summary.append({
        "file": file.name,
        "original_rows": len(df),
        "cleaned_rows": len(cleaned_df)
    })

cleaning_summary_df = pd.DataFrame(cleaning_summary)

print(
    f"Cleaning complete. {len(cleaning_summary_df)} files processed."
)

display(cleaning_summary_df)
```

```python
cleaned_files = sorted(interim_folder.glob("*.csv"))

print("Number of cleaned files:", len(cleaned_files))

missing_cleaned_files = [
    file.name
    for file in raw_files
    if not (interim_folder / file.name).exists()
]

print("Missing cleaned files:", len(missing_cleaned_files))

if missing_cleaned_files:
    print(missing_cleaned_files)
```

```python
for file in cleaned_files[:3]:

    df = pd.read_csv(file)

    print("\nFile:", file.name)
    print(df.dtypes)
```

```python
comma_results = []

for file in cleaned_files:

    df = pd.read_csv(file)

    commas = (
        df["registered_voters"]
        .astype(str)
        .str.contains(",", regex=False)
        .sum()
    )

    comma_results.append({
        "file": file.name,
        "registered_voters_with_commas": commas
    })

comma_df = pd.DataFrame(comma_results)

display(comma_df)

print(
    "Total registered voter values still containing commas:",
    comma_df["registered_voters_with_commas"].sum()
)
```

```python
cleaned_missing = []

for file in cleaned_files:

    df = pd.read_csv(file)

    cleaned_missing.append({
        "file": file.name,
        "missing_values": df.isnull().sum().sum()
    })

cleaned_missing_df = pd.DataFrame(cleaned_missing)

display(cleaned_missing_df)

print(
    "Total missing values after cleaning:",
    cleaned_missing_df["missing_values"].sum()
)
```

```python
cleaned_duplicates = []

for file in cleaned_files:

    df = pd.read_csv(file)

    cleaned_duplicates.append({
        "file": file.name,
        "duplicate_wards": df["ward"].duplicated().sum()
    })

cleaned_duplicates_df = pd.DataFrame(cleaned_duplicates)

display(cleaned_duplicates_df)

print(
    "Total duplicate wards after cleaning:",
    cleaned_duplicates_df["duplicate_wards"].sum()
)
```

```python
range_results = []

for file in cleaned_files:

    df = pd.read_csv(file)

    invalid_registered = (df["registered_voters"] < 0).sum()
    invalid_districts = (df["voting_districts"] <= 0).sum()

    range_results.append({
        "file": file.name,
        "negative_registered_voters": invalid_registered,
        "invalid_voting_districts": invalid_districts
    })

range_df = pd.DataFrame(range_results)

display(range_df)

print(
    "Negative registered voter values:",
    range_df["negative_registered_voters"].sum()
)

print(
    "Invalid voting district values:",
    range_df["invalid_voting_districts"].sum()
)
```

```python
row_check = []

for raw_file in raw_files:

    cleaned_file = interim_folder / raw_file.name

    raw_df = pd.read_csv(raw_file)
    cleaned_df = pd.read_csv(cleaned_file)

    row_check.append({
        "file": raw_file.name,
        "raw_rows": len(raw_df),
        "cleaned_rows": len(cleaned_df),
        "rows_preserved": len(raw_df) == len(cleaned_df)
    })

row_check_df = pd.DataFrame(row_check)

display(row_check_df)

print(
    "Files where row counts were not preserved:",
    (~row_check_df["rows_preserved"]).sum()
)
```

```python
value_check = []

for raw_file in raw_files:

    cleaned_file = interim_folder / raw_file.name

    raw_df = pd.read_csv(raw_file)
    cleaned_df = pd.read_csv(cleaned_file)

    raw_values = (
        raw_df["registered_voters"]
        .astype(str)
        .str.strip()
        .str.replace(",", "", regex=False)
    )

    cleaned_values = (
        cleaned_df["registered_voters"]
        .astype(str)
        .str.strip()
    )

    mismatches = (
        raw_values != cleaned_values
    ).sum()

    value_check.append({
        "file": raw_file.name,
        "registered_voter_value_mismatches": mismatches
    })

value_check_df = pd.DataFrame(value_check)

display(value_check_df)

print(
    "Total registered voter value mismatches:",
    value_check_df["registered_voter_value_mismatches"].sum()
)
```

```python
all_cleaned_data = []

for file in cleaned_files:

    df = pd.read_csv(file)

    df["source_file"] = file.name

    all_cleaned_data.append(df)

combined_check = pd.concat(
    all_cleaned_data,
    ignore_index=True
)

print("Total ward records:", len(combined_check))
print(
    "Minimum registered voters in a ward:",
    combined_check["registered_voters"].min()
)
print(
    "Maximum registered voters in a ward:",
    combined_check["registered_voters"].max()
)
print(
    "Wards with zero registered voters:",
    (combined_check["registered_voters"] == 0).sum()
)
```

```python
duplicate_wards_across_files = (
    combined_check["ward"]
    .duplicated()
    .sum()
)

print(
    "Duplicate ward codes across all 44 files:",
    duplicate_wards_across_files
)
```

```python
print("=" * 60)
print("FINAL CLEANING VALIDATION")
print("=" * 60)

print("Raw files:", len(raw_files))
print("Cleaned files:", len(cleaned_files))
print("Total ward records:", len(combined_check))

print(
    "Total missing values:",
    cleaned_missing_df["missing_values"].sum()
)

print(
    "Duplicate wards within files:",
    cleaned_duplicates_df["duplicate_wards"].sum()
)

print(
    "Duplicate ward codes across files:",
    duplicate_wards_across_files
)

print(
    "Registered voter values with commas:",
    comma_df["registered_voters_with_commas"].sum()
)

print(
    "Registered voter value mismatches:",
    value_check_df["registered_voter_value_mismatches"].sum()
)

print(
    "Files with lost/added rows:",
    (~row_check_df["rows_preserved"]).sum()
)

print(
    "Cleaned folder exists:",
    interim_folder.exists()
)

print(
    "All cleaned files exist:",
    len(cleaned_files) == len(raw_files)
)
```
