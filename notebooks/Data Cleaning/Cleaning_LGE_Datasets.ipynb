```python
#Okay tric here would be sure to locate files and be sure of where are and then inspect them 
# lets do that then what we get will determine our game plan

from pathlib import Path
import pandas as pd


RAW_DIR = Path(
    r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA"
    r"\data\raw\KZN_Local_Goverment_elections_2011-21"
)

files = [
    RAW_DIR / "KZN_2011.csv",
    RAW_DIR / "KZN_2016.csv",
    RAW_DIR / "KZN_2021.csv",
]


# ---------------------------------------------------------
# ENCODING DETECTION
# ---------------------------------------------------------

def detect_encoding(file_path):
    """
    Detect common encodings using BOM first,
    then try common Windows/Unicode encodings.
    """
    
    with open(file_path, "rb") as f:
        raw = f.read(4)

    # Byte Order Marks
    if raw.startswith(b"\xff\xfe"):
        return "utf-16"
    
    if raw.startswith(b"\xfe\xff"):
        return "utf-16"
    
    if raw.startswith(b"\xef\xbb\xbf"):
        return "utf-8-sig"

    # No BOM: test common encodings
    for encoding in ["utf-8", "cp1252", "latin1"]:
        try:
            with open(file_path, "r", encoding=encoding) as f:
                f.read(5000)
            return encoding
        except UnicodeDecodeError:
            continue

    return "unknown"


# ---------------------------------------------------------
# INSPECT EACH FILE
# ---------------------------------------------------------

for file in files:

    print("=" * 90)
    print(f"FILE: {file.name}")
    print("=" * 90)

    if not file.exists():
        print("WARNING: File not found")
        continue

    encoding = detect_encoding(file)

    print(f"Detected encoding: {encoding}")

    if encoding == "unknown":
        print("Could not determine encoding safely.")
        continue

    # Try reading the CSV
    try:
        df = pd.read_csv(
            file,
            encoding=encoding
        )

        print(f"Rows: {len(df):,}")
        print(f"Columns: {len(df.columns)}")

        print("\nColumns:")
        print(df.columns.tolist())

        print("\nFirst 5 rows:")
        display(df.head())

        print("\nData types:")
        print(df.dtypes)

        print("\nMissing values:")
        print(df.isna().sum())

        print("\nDuplicate rows:")
        print(df.duplicated().sum())

    except Exception as e:
        print(f"READ ERROR: {type(e).__name__}: {e}")

    print()
```

```python
from pathlib import Path
import pandas as pd

# Set the raw and cleaned data folders
BASE_DIR = Path(r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA")
RAW_DIR = BASE_DIR / "data" / "raw" / "KZN_Local_Goverment_elections_2011-21"
CLEAN_DIR = BASE_DIR / "data" / "interim" / "KZN_Local_Goverment_elections_2011-21 Cleaned"

CLEAN_DIR.mkdir(parents=True, exist_ok=True)

# Read each election year using its detected encoding
files = {
    2011: ("KZN_2011.csv", "utf-16"),
    2016: ("KZN_2016.csv", "utf-8-sig"),
    2021: ("KZN_2021.csv", "utf-8-sig")
}

data = {}

for year, (filename, encoding) in files.items():
    data[year] = pd.read_csv(
        RAW_DIR / filename,
        encoding=encoding,
        low_memory=False
    )
    print(f"{year}: {len(data[year]):,} rows loaded")
```

```python
# Okay now lets begin our data cleaning


# Clean column names and text fields without changing their meaning
# This keeps party, municipality, station and ward values consistent

text_columns = [
    "Province",
    "Municipality",
    "Ward",
    "VotingStationName",
    "BallotType",
    "PartyName"
]

for year, df in data.items():

    # Clean column names
    df.columns = (
        df.columns
        .str.replace("\ufeff", "", regex=False)
        .str.strip()
    )

    # Clean text values
    for column in text_columns:
        df[column] = (
            df[column]
            .astype("string")
            .str.replace("\u00a0", " ", regex=False)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

    print(f"\n{year}")
    print("Columns:", df.columns.tolist())
    print("Text fields cleaned successfully")
```

```python
# Check election counts for invalid, negative or missing values
# We keep the original values untouched until the checks are complete

numeric_columns = [
    "RegisteredVoters",
    "SpoiltVotes",
    "TotalValidVotes"
]

for year, df in data.items():

    print("\n" + "=" * 70)
    print(f"KZN {year} — NUMERIC DATA CHECK")
    print("=" * 70)

    for column in numeric_columns:

        # Convert only for checking; the dataset is not changed here
        values = pd.to_numeric(df[column], errors="coerce")

        print(f"\n{column}")
        print(f"  Missing: {values.isna().sum():,}")
        print(f"  Negative: {(values < 0).sum():,}")
        print(f"  Minimum: {values.min()}")
        print(f"  Maximum: {values.max()}")

    # Check whether valid votes exceed registered voters
    registered = pd.to_numeric(
        df["RegisteredVoters"],
        errors="coerce"
    )

    valid = pd.to_numeric(
        df["TotalValidVotes"],
        errors="coerce"
    )

    over_registered = valid > registered

    print(
        f"\nValid votes greater than registered voters: "
        f"{over_registered.sum():,}"
    )
```

```python
# Inspect records where valid votes exceed registered voters
# These records will not be deleted until we understand why they occur

for year, df in data.items():

    registered = pd.to_numeric(
        df["RegisteredVoters"],
        errors="coerce"
    )

    valid = pd.to_numeric(
        df["TotalValidVotes"],
        errors="coerce"
    )

    suspicious = df.loc[
        valid > registered,
        [
            "Province",
            "Municipality",
            "Ward",
            "VotingDistrict",
            "VotingStationName",
            "RegisteredVoters",
            "SpoiltVotes",
            "PartyName",
            "TotalValidVotes"
        ]
    ]

    print("\n" + "=" * 90)
    print(f"KZN {year} — SUSPICIOUS VOTE RECORDS")
    print("=" * 90)
    print(f"Records found: {len(suspicious):,}")

    display(suspicious)
```

```python
# Check for repeated electoral records within each election year
# Duplicates are reviewed before any records are removed

duplicate_columns = [
    "Municipality",
    "Ward",
    "VotingDistrict",
    "VotingStationName",
    "BallotType",
    "PartyName",
    "TotalValidVotes"
]

for year, df in data.items():

    duplicates = df[df.duplicated(
        subset=duplicate_columns,
        keep=False
    )].sort_values(duplicate_columns)

    print("\n" + "=" * 80)
    print(f"KZN {year} — DUPLICATE RECORD CHECK")
    print("=" * 80)
    print(f"Duplicate records found: {len(duplicates):,}")

    if len(duplicates) > 0:
        display(duplicates.head(30))
```

```python
# Check vote-count columns for formatting and data-type issues
# This identifies values that need conversion before the final clean save

vote_columns = [
    "RegisteredVoters",
    "SpoiltVotes",
    "TotalValidVotes"
]

for year, df in data.items():

    print("\n" + "=" * 80)
    print(f"KZN {year} — VOTE COUNT FORMAT CHECK")
    print("=" * 80)

    for column in vote_columns:

        numeric_check = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        print(f"\n{column}")
        print(f"  Original type: {df[column].dtype}")
        print(f"  Non-numeric values: {numeric_check.isna().sum():,}")
        print(f"  Unique values: {df[column].nunique():,}")
```

```python
# Check ward-code formatting and leading-zero preservation
# Ward values are inspected before any formatting changes are made

for year, df in data.items():

    print("\n" + "=" * 80)
    print(f"KZN {year} — WARD CODE FORMAT CHECK")
    print("=" * 80)

    ward_values = df["Ward"].astype("string").str.strip()

    print(f"Unique ward values: {ward_values.nunique():,}")
    print(f"Missing ward values: {ward_values.isna().sum():,}")

    # Check ward values that do not follow the expected "Ward XXXXXXXX" format
    invalid_wards = ward_values[
        ~ward_values.str.match(r"^Ward \d+$", na=False)
    ]

    print(f"Unexpected ward formats: {len(invalid_wards):,}")

    if len(invalid_wards) > 0:
        display(
            pd.DataFrame({
                "Ward": invalid_wards.drop_duplicates()
            }).head(30)
        )
```

```python
# Check party-name consistency within each election year
# This identifies spelling and formatting variations before standardization

for year, df in data.items():

    print("\n" + "=" * 80)
    print(f"KZN {year} — PARTY NAME CHECK")
    print("=" * 80)

    party_names = (
        df["PartyName"]
        .dropna()
        .astype("string")
        .str.strip()
    )

    print(f"Unique party names: {party_names.nunique():,}")

    display(
        pd.DataFrame({
            "PartyName": sorted(party_names.unique())
        })
    )
```

```python
# Check for hidden party-name variations after text normalization
# Genuine historical party names are kept unchanged

for year, df in data.items():

    normalized = (
        df["PartyName"]
        .astype("string")
        .str.upper()
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    check = pd.DataFrame({
        "Original": df["PartyName"],
        "Normalized": normalized
    }).drop_duplicates()

    variations = check[
        check.duplicated("Normalized", keep=False)
    ].sort_values("Normalized")

    print("\n" + "=" * 80)
    print(f"KZN {year} — NORMALIZED PARTY NAME CHECK")
    print("=" * 80)
    print(f"Hidden variations found: {len(variations):,}")

    if len(variations) > 0:
        display(variations)
        
```

```python
# Apply confirmed cleaning changes and save the interim datasets
# Raw files remain untouched and only cleaned copies are written

for year, df in data.items():

    clean_df = df.copy()

    # Standardize text fields without changing their meaning
    for column in text_columns:
        clean_df[column] = (
            clean_df[column]
            .astype("string")
            .str.replace("\u00a0", " ", regex=False)
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

    # Keep vote-count columns as numeric integers
    for column in numeric_columns:
        clean_df[column] = pd.to_numeric(
            clean_df[column],
            errors="raise"
        ).astype("int64")

    # Validate the cleaned dataset before saving
    assert clean_df.columns.tolist() == [
        "Province",
        "Municipality",
        "Ward",
        "VotingDistrict",
        "VotingStationName",
        "RegisteredVoters",
        "BallotType",
        "SpoiltVotes",
        "PartyName",
        "TotalValidVotes",
        "DateGenerated"
    ]

    assert clean_df.isna().sum().sum() == 0
    assert clean_df.duplicated().sum() == 0

    output_file = CLEAN_DIR / f"KZN_{year}_clean.csv"

    clean_df.to_csv(
        output_file,
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"KZN {year}: {len(clean_df):,} rows saved -> "
        f"{output_file.name}"
    )
```
