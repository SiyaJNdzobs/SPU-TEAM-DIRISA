```python
from pathlib import Path
import pandas as pd
import requests
from bs4 import BeautifulSoup

PROJECT_ROOT = Path.cwd().parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

AFROBAROMETER_RAW_DIR = RAW_DIR / "afrobarometer"
AFROBAROMETER_PROCESSED_DIR = PROCESSED_DIR / "afrobarometer"

AFROBAROMETER_RAW_DIR.mkdir(parents=True, exist_ok=True)
AFROBAROMETER_PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw directory:", AFROBAROMETER_RAW_DIR)
print("Processed directory:", AFROBAROMETER_PROCESSED_DIR)
```

```python
AFROBAROMETER_URL = "https://www.afrobarometer.org/articles/a-long-to-do-list-for-the-victors-of-south-africas-2024-election/"

print("Source:", AFROBAROMETER_URL)
```

```python
response = requests.get(
    AFROBAROMETER_URL,
    timeout=30
)

print("Status code:", response.status_code)
print("Page size:", len(response.text), "characters")
```

```python
soup = BeautifulSoup(response.text, "html.parser")

title = soup.title.get_text(strip=True) if soup.title else None
headings = [h.get_text(" ", strip=True) for h in soup.find_all(["h1", "h2", "h3"])]
paragraphs = [p.get_text(" ", strip=True) for p in soup.find_all("p")]

print("PAGE TITLE:")
print(title)

print("\nNUMBER OF HEADINGS:", len(headings))
print("NUMBER OF PARAGRAPHS:", len(paragraphs))

print("\nHEADINGS:")
for heading in headings:
    print("-", heading)
```

```python
for i, paragraph in enumerate(paragraphs):
    print(f"[{i}] {paragraph}")
```

```python
for i in range(24, 42):
    print(f"[{i}] {paragraphs[i]}")
```

```python
tables = soup.find_all("table")

print("Number of HTML tables:", len(tables))

images = soup.find_all("img")

print("Number of images:", len(images))

for image in images:
    alt = image.get("alt")
    if alt:
        print("-", alt)
```

```python
for i, image in enumerate(images):
    print(f"\nIMAGE {i + 1}")
    print("Alt:", image.get("alt"))
    print("Source:", image.get("src"))
```

```python
links = soup.find_all("a")

for link in links:
    text = link.get_text(" ", strip=True)
    href = link.get("href")

    if text or href:
        print("TEXT:", text)
        print("LINK:", href)
        print()
```

```python
for link in links:
    text = link.get_text(" ", strip=True).lower()
    href = link.get("href")

    if href and any(word in text + " " + href.lower()
                    for word in ["data", "dataset", "download", "analyse", "analysis"]):
        print("TEXT:", link.get_text(" ", strip=True))
        print("LINK:", href)
        print()
```

```python
kzn_data = sav_data[sav_data["REGION"] == 703.0].copy()

print("KZN rows:", kzn_data.shape[0])
print("KZN columns:", kzn_data.shape[1])
```

```python
selected_variables = [
    "Q13",
    "Q9C",
    "Q12A",
    "Q12B",
    "Q14A",
    "Q14B",
    "Q14C",
    "Q19B",
    "Q24",
    "Q23B_SAF",
    "URBRUR",
    "Q3",
    "Q4A",
    "Q4B"
]

missing_variables = [
    variable for variable in selected_variables
    if variable not in kzn_data.columns
]

print("Selected variables:", len(selected_variables))
print("Missing variables:", missing_variables)
```

```python
kzn_voter_data = kzn_data[selected_variables].copy()

print("KZN voter dataset shape:", kzn_voter_data.shape)
print("\nFirst 5 rows:")
print(kzn_voter_data.head())
```

```python
KZN_AFROBAROMETER_FILE = (
    AFROBAROMETER_PROCESSED_DIR
    / "kzn_afrobarometer_round9_voter_data.csv"
)

kzn_voter_data.to_csv(
    KZN_AFROBAROMETER_FILE,
    index=False
)

print("Saved successfully:")
print(KZN_AFROBAROMETER_FILE)
print("\nFile exists:", KZN_AFROBAROMETER_FILE.exists())
```

```python
for variable in selected_variables:
    print("\n" + "=" * 70)
    print(variable)
    print("Question:", sav_metadata.column_names_to_labels.get(variable))

    labels = sav_metadata.variable_value_labels.get(variable, {})
    print("Response codes:")

    for code, label in labels.items():
        print(f"  {code} = {label}")
```

```python
# Create a readable copy of the KZN dataset
kzn_readable_data = kzn_voter_data.copy()

# Rename columns using the official Afrobarometer question labels
readable_column_names = {
    variable: sav_metadata.column_names_to_labels.get(variable, variable)
    for variable in selected_variables
}

kzn_readable_data = kzn_readable_data.rename(
    columns=readable_column_names
)

# Convert the numerical response codes into their official labels
for variable in selected_variables:
    original_question = sav_metadata.column_names_to_labels.get(
        variable, variable
    )
    
    value_labels = sav_metadata.variable_value_labels.get(variable, {})
    
    if value_labels:
        kzn_readable_data[original_question] = (
            kzn_readable_data[original_question]
            .map(value_labels)
            .fillna(kzn_readable_data[original_question])
        )

print("Readable KZN dataset shape:", kzn_readable_data.shape)
print("\nColumns:")
for column in kzn_readable_data.columns:
    print("-", column)
```

```python
# Fix encoding issues in the readable Afrobarometer dataset

for column in kzn_readable_data.columns:
    if kzn_readable_data[column].dtype == "object":
        kzn_readable_data[column] = (
            kzn_readable_data[column]
            .astype(str)
            .str.replace("â€™", "'", regex=False)
            .str.replace("â€˜", "'", regex=False)
            .str.replace("â€œ", '"', regex=False)
            .str.replace("â€", '"', regex=False)
        )

# Save the corrected readable dataset
kzn_readable_data.to_csv(
    AFROBAROMETER_PROCESSED_DIR / "kzn_afrobarometer_round9_readable_voter_data.csv",
    index=False,
    encoding="utf-8"
)

print("Encoding issues fixed and dataset saved.")
```
