```python

#1.  Setup: KZN Water Access Dataset


from pathlib import Path
import requests
import pandas as pd
from bs4 import BeautifulSoup

# Project root
PROJECT_ROOT = Path(
    r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA"
)

# Dedicated raw-data folder for this dataset
WATER_RAW_DIR = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "kzn_water_access_indicators"
)

# Create folder 
WATER_RAW_DIR.mkdir(parents=True, exist_ok=True)

# Africa Check source
WATER_URL = (
    "https://africacheck.org/fact-checks/"
    "factsheets/election-factsheet-water-south-africa"
)

# Output file
WATER_OUTPUT = WATER_RAW_DIR / "kzn_water_access_indicators.csv"

print("Setup completed successfully.")
print(f"Project root: {PROJECT_ROOT}")
print(f"Dataset folder: {WATER_RAW_DIR}")
print(f"Source: {WATER_URL}")
print(f"Output file: {WATER_OUTPUT}")

# Basic verification
assert PROJECT_ROOT.exists(), "Project root does not exist."
assert WATER_RAW_DIR.exists(), "Dataset folder could not be created."

print("Path verification: PASSED")
```

```python
#2.

#Retrieve and Verify Africa Check Source
# Purpose:
# Confirm that the Africa Check water factsheet is accessible
# and contains the KZN water-access figures required for the
# raw dataset.


response = requests.get(
    WATER_URL,
    headers={
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/153.0.0.0 Safari/537.36"
        )
    },
    timeout=30
)

print(f"HTTP status: {response.status_code}")
print(f"Content type: {response.headers.get('Content-Type')}")
print(f"Response size: {len(response.content)} bytes")

# Verify successful response
assert response.status_code == 200, (
    f"Source request failed: HTTP {response.status_code}"
)

# Parse page
soup = BeautifulSoup(response.text, "html.parser")
page_text = soup.get_text(" ", strip=True)

# Verify the KZN-specific figures reported by the source
assert "KwaZulu-Natal" in page_text
assert "75.4%" in page_text
assert "81.5%" in page_text
assert "2002" in page_text
assert "2023" in page_text

print()
print("Source retrieval: PASSED")
print("KZN water-access figures detected:")
print("2002: 75.4%")
print("2023: 81.5%")
print("Source verification: PASSED")
```
