### Analytical Audit Across 5 Election Cycles
This notebook audits crosswalk coverage, turnout distributions across 2000-2021, missingness, and municipal participation shifts.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = Path.cwd().parent.parent if Path.cwd().name == 'EDA and Visualization' else Path.cwd()
PROCESSED_DIR = BASE_DIR / 'data' / 'processed'

df = pd.read_csv(PROCESSED_DIR / '08_feature_engineered_dataset' / 'ward_features_2000_2021_crosswalk_corrected.csv')
print('Loaded dataset shape:', df.shape)
print('Yearly distribution:\n', df['ElectionYear'].value_counts().sort_index())
```

### Audit 1: Longitudinal Turnout Trends Across KZN
We inspect provincial mean and median turnout across all five municipal elections.

```python
summary = df.groupby('ElectionYear')['TurnoutRate'].agg(['count', 'mean', 'std', 'min', 'median', 'max']).reset_index()
print(summary.to_string(index=False))

plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='ElectionYear', y='TurnoutRate', palette='Blues')
plt.title('Ward Turnout Rate Distribution Across 5 KZN Elections (2000–2021)', fontsize=13, fontweight='bold')
plt.xlabel('Election Cycle')
plt.ylabel('Turnout Rate (%)')
plt.tight_layout()
plt.show()
```
