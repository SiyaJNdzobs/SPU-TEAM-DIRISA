```python
from pathlib import Path
import pandas as pd

# Set the cleaned Afrobarometer source and processed output paths.
# We keep this survey data separate because it provides supporting political-attitudes context, not ward coverage.

source_path = Path(
    r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA\data\interim\KZN_Voting_Political_Attitudes_Afrobarometer-2024 Cleaned\kzn_afrobarometer_round9_readable_voter_data_clean.csv"
)

output_dir = Path(
    r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA\data\processed\06_afrobarometer_supporting"
)

output_dir.mkdir(parents=True, exist_ok=True)

print("Source exists:", source_path.exists())
print("Output folder:", output_dir)
```

```python
# Load the Afrobarometer survey using its semicolon-separated structure.
# This gives us the respondent-level political-attitudes data for supporting analysis.

afrobarometer_df = pd.read_csv(
    source_path,
    sep=";",
    encoding="utf-8-sig"
)

print("Shape:", afrobarometer_df.shape)
print("\nColumns:")
print(afrobarometer_df.columns.tolist())

print("\nFirst 5 rows:")
display(afrobarometer_df.head())
```

```python
# Inspect the first lines of the Afrobarometer file to identify its actual CSV structure.
# We need to read the survey correctly before using it as supporting political-attitudes context.

with open(source_path, "r", encoding="utf-8-sig", errors="replace") as file:
    for i in range(12):
        print(f"{i + 1}: {file.readline().rstrip()}")
```

```python
# Check the survey sample structure and missing responses before saving it as supporting data.
# This confirms what the 248 respondents can contribute to the turnout project without forcing a ward-level merge.

print("Respondents:", len(afrobarometer_df))
print("Variables:", len(afrobarometer_df.columns))

print("\nUrban/Rural sample:")
display(
    afrobarometer_df["Urban or Rural Primary Sampling Unit"]
    .value_counts(dropna=False)
)

print("\nMissing values by variable:")
display(
    afrobarometer_df.isna().sum()
)
```

```python
# Confirm that each row represents one survey respondent and that no rows are duplicated.
# This keeps the Afrobarometer data at its original survey grain for supporting analysis.

duplicate_rows = afrobarometer_df.duplicated().sum()

print("Respondents:", len(afrobarometer_df))
print("Duplicate complete rows:", duplicate_rows)

if duplicate_rows == 0:
    print("\nSurvey grain validation: PASSED")
    print("Decision: Keep as respondent-level supporting data.")
else:
    print("\nSurvey grain validation: REVIEW REQUIRED")
```

```python
# Save the validated Afrobarometer survey as supporting data for later analysis.
# We keep the original respondent-level structure so its survey evidence is not distorted by a ward-level merge.

output_path = output_dir / "kzn_afrobarometer_political_attitudes_2024.csv"

afrobarometer_df.to_csv(
    output_path,
    index=False,
    sep=";"
)

print("Afrobarometer supporting dataset saved.")
print("Output file:", output_path)
print("Rows saved:", len(afrobarometer_df))
print("Columns saved:", len(afrobarometer_df.columns))
```

```python
# Re-open the saved survey and confirm the processed file still contains all respondents and variables.
# This final check confirms the supporting dataset is ready for the project pipeline.

saved_afrobarometer = pd.read_csv(
    output_path,
    sep=";",
    encoding="utf-8-sig"
)

print("File exists:", output_path.exists())
print("Rows loaded:", len(saved_afrobarometer))
print("Columns loaded:", len(saved_afrobarometer.columns))

if (
    output_path.exists()
    and len(saved_afrobarometer) == 248
    and len(saved_afrobarometer.columns) == 14
):
    print("\nAfrobarometer save validation: PASSED")
else:
    print("\nAfrobarometer save validation: FAILED")
```
