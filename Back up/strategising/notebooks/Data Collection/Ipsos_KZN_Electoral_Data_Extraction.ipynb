```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")

article_text = soup.get_text(
    separator="\n",
    strip=True
)

print("Article text length:", len(article_text))

print("\n--- KZN / PROVINCE ---")
for keyword in ["KwaZulu-Natal", "KZN", "Gauteng", "Northern Cape", "North West", "Limpopo"]:
    if keyword.lower() in article_text.lower():
        print(f"Found: {keyword}")

print("\n--- AGE ---")
for keyword in ["18-24", "25-34", "18–24", "25–34", "young"]:
    if keyword.lower() in article_text.lower():
        print(f"Found: {keyword}")

print("\n--- TRUST ---")
for keyword in ["trust", "IEC", "electoral"]:
    print(f"{keyword}: {article_text.lower().count(keyword.lower())} occurrence(s)")
```

```python
keywords = [
    "KwaZulu-Natal",
    "KZN",
    "18-24",
    "25-34",
    "trust",
    "registered",
    "vote",
    "politics"
]

for keyword in keywords:
    print("\n" + "=" * 80)
    print(f"SEARCHING FOR: {keyword}")
    print("=" * 80)

    text_lower = article_text.lower()
    keyword_lower = keyword.lower()

    start = 0
    found = False

    while True:
        position = text_lower.find(keyword_lower, start)

        if position == -1:
            break

        found = True

        # Show surrounding text
        section_start = max(0, position - 350)
        section_end = min(len(article_text), position + 700)

        print(article_text[section_start:section_end])
        print("-" * 80)

        start = position + len(keyword_lower)

    if not found:
        print("No matches found.")
```
