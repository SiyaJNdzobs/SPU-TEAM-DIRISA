```python
#2. — Load and Inspect Province Context Data
# Purpose:
# Load the socioeconomic, youth distrust and QLFS datasets so we can
# confirm their structure before building the province context panel.

context_data = {}

for name, file_path in CONTEXT_FILES.items():
    df = pd.read_csv(file_path)
    context_data[name] = df

    print(f"{name} dataset loaded successfully.")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")
    print(f"Columns: {df.columns.tolist()}")
    print()

assert set(context_data.keys()) == set(CONTEXT_FILES.keys())

print("All province context datasets loaded successfully.")
print("Datasets available:", sorted(context_data.keys()))
```

```python
#3. — Inspect Province Context Values
# Purpose:
# Inspect the actual geography, year and indicator values so we can
# identify which records can safely contribute to the KZN context panel.

for name, df in context_data.items():

    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    if name == "youth_distrust":
        print("Geography values:")
        print(df["geography"].unique())
        print()
        print("Years:")
        print(df["year"].unique())
        print()
        print("Indicators:")
        print(df["indicator"].unique())

    elif name == "socioeconomic":
        print("Geography levels:")
        print(df["geography_level"].unique())
        print()
        print("Geography names:")
        print(df["geography_name"].unique())
        print()
        print("Years:")
        print(sorted(df["year"].dropna().unique()))
        print()
        print("Indicators:")
        print(df["indicator"].unique())

    elif name == "qlfs_q1_2024":
        print("Geographies:")
        print(df["Geography"].unique())
        print()
        print("Metrics:")
        print(df["Metric"].unique())
        print()

    elif name == "qlfs_2024":
        print("Provinces:")
        print(df["Province"].unique())
        print()
        print("Institution types:")
        print(df["Institution_type"].unique())
        print()

    elif name == "qlfs_q1_2026":
        print("Geographies:")
        print(df["Geography"].unique())
        print()
        print("Periods:")
        print(df["Period"].unique())
        print()
        print("Indicators:")
        print(df["Indicator"].unique())

    print()
```

```python
#4. — Inspect Usable KZN Records
# Purpose:
# Inspect the KZN province records that may enter the context panel
# before deciding how each dataset should be matched by year.

youth_df = context_data["youth_distrust"]
socioeconomic_df = context_data["socioeconomic"]
qlfs_q1_2024_df = context_data["qlfs_q1_2024"]
qlfs_2024_df = context_data["qlfs_2024"]
qlfs_q1_2026_df = context_data["qlfs_q1_2026"]

print("=" * 70)
print("YOUTH DISTRUST — KZN 2023")
print("=" * 70)
display(youth_df)

print()
print("=" * 70)
print("SOCIOECONOMIC — KZN PROVINCE ONLY")
print("=" * 70)

kzn_socioeconomic = socioeconomic_df[
    socioeconomic_df["geography_name"] == "KwaZulu-Natal"
].copy()

display(kzn_socioeconomic)

print()
print("=" * 70)
print("QLFS Q1 2024 — KZN PROVINCE ONLY")
print("=" * 70)

kzn_qlfs_q1_2024 = qlfs_q1_2024_df[
    qlfs_q1_2024_df["Geography"] == "KwaZulu-Natal"
].copy()

display(kzn_qlfs_q1_2024)

print()
print("=" * 70)
print("QLFS 2024 — KZN")
print("=" * 70)

display(qlfs_2024_df)

print()
print("=" * 70)
print("QLFS Q1 2026 — NATIONAL DATA")
print("=" * 70)

display(qlfs_q1_2026_df)

print()
print("KZN province-record inspection completed.")
```

```python
#5. — Prepare Province-Level Sources
# Purpose:
# Keep only KZN province records and identify the time coverage
# of each source before building the province context panel.

province_sources = {}

# Socioeconomic indicators — KZN province only
province_sources["socioeconomic"] = (
    socioeconomic_df[
        (socioeconomic_df["geography_level"] == "Province")
        & (socioeconomic_df["geography_name"] == "KwaZulu-Natal")
    ]
    .copy()
)

# Youth distrust — KZN only
province_sources["youth_distrust"] = (
    youth_df[
        youth_df["geography"] == "KwaZulu-Natal"
    ]
    .copy()
)

# QLFS Q1 2024 — KZN province only
province_sources["qlfs_q1_2024"] = (
    qlfs_q1_2024_df[
        qlfs_q1_2024_df["Geography"] == "KwaZulu-Natal"
    ]
    .copy()
)

for name, df in province_sources.items():
    print(f"{name}:")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")

    if name == "socioeconomic":
        print(f"Years: {sorted(df['year'].unique())}")

    elif name == "youth_distrust":
        print(f"Years: {sorted(df['year'].unique())}")

    elif name == "qlfs_q1_2024":
        print(f"Periods available: Q1 2023 to Q1 2024")

    print()

assert len(province_sources["socioeconomic"]) == 18
assert len(province_sources["youth_distrust"]) == 5
assert len(province_sources["qlfs_q1_2024"]) == 10

print("Province-level source preparation: PASSED")
```

```python
#6. — Build Socioeconomic Province-Year Panel
# Purpose:
# Reshape the KZN socioeconomic indicators into one row per year
# so the province context panel has a consistent time structure.

socioeconomic_panel = (
    province_sources["socioeconomic"]
    .pivot_table(
        index=["geography_level", "geography_name", "year"],
        columns="indicator",
        values="value",
        aggfunc="first"
    )
    .reset_index()
)

socioeconomic_panel.columns.name = None

print("Socioeconomic province-year panel created.")
print(f"Rows: {len(socioeconomic_panel):,}")
print(f"Columns: {len(socioeconomic_panel.columns)}")
print()

print("Years:")
print(sorted(socioeconomic_panel["year"].unique()))

print()
print("Columns:")
print(socioeconomic_panel.columns.tolist())

print()
print("Panel preview:")
display(socioeconomic_panel)

assert len(socioeconomic_panel) == 3
assert socioeconomic_panel["year"].nunique() == 3
assert socioeconomic_panel["geography_name"].nunique() == 1

print()
print("Socioeconomic province-year panel: PASSED")
```

```python
#7. — Reshape Youth Distrust Indicators
# Purpose:
# Reshape the 2023 KZN youth distrust indicators into columns
# so they can be attached to the matching province-year record.

youth_distrust_panel = (
    province_sources["youth_distrust"]
    .pivot_table(
        index=["geography", "year"],
        columns="indicator",
        values="value",
        aggfunc="first"
    )
    .reset_index()
)

youth_distrust_panel.columns.name = None

print("Youth distrust panel created.")
print(f"Rows: {len(youth_distrust_panel):,}")
print(f"Columns: {len(youth_distrust_panel.columns)}")
print()

print("Columns:")
print(youth_distrust_panel.columns.tolist())

print()
print("Panel preview:")
display(youth_distrust_panel)

assert len(youth_distrust_panel) == 1
assert youth_distrust_panel["year"].iloc[0] == 2023
assert youth_distrust_panel["geography"].iloc[0] == "KwaZulu-Natal"

print()
print("Youth distrust reshaping: PASSED")
```

```python
#8. — Merge Youth Distrust with Socioeconomic Panel
# Purpose:
# Add the 2023 youth distrust indicators to the matching KZN
# province-year record without creating duplicate province-year rows.

province_context_panel = socioeconomic_panel.merge(
    youth_distrust_panel,
    left_on=["geography_name", "year"],
    right_on=["geography", "year"],
    how="left",
    validate="one_to_one"
)

province_context_panel = province_context_panel.drop(
    columns=["geography"]
)

print("Youth distrust merged successfully.")
print(f"Rows: {len(province_context_panel):,}")
print(f"Columns: {len(province_context_panel.columns)}")
print()

print("Years:")
print(sorted(province_context_panel["year"].unique()))

print()
print("Youth distrust availability:")
print(
    province_context_panel[
        [
            "year",
            "Dissatisfied with democracy",
            "Distrust of national government",
            "Distrust of political parties",
            "Distrust of provincial government",
            "Distrust of the IEC"
        ]
    ]
)

assert len(province_context_panel) == 3
assert province_context_panel["year"].nunique() == 3

print()
print("Youth distrust merge: PASSED")
```

```python
#9. — Add QLFS Q1 2024 Context
# Purpose:
# Add the KZN Q1 2024 labour-market indicators as a dated province
# context snapshot without assigning them to earlier years.

qlfs_2024_context = (
    province_sources["qlfs_q1_2024"]
    [["Geography", "Metric", "Q1_2024"]]
    .copy()
)

qlfs_2024_context = (
    qlfs_2024_context
    .pivot_table(
        index="Geography",
        columns="Metric",
        values="Q1_2024",
        aggfunc="first"
    )
    .reset_index()
)

qlfs_2024_context.columns.name = None

print("QLFS Q1 2024 context created.")
print(f"Rows: {len(qlfs_2024_context):,}")
print(f"Columns: {len(qlfs_2024_context.columns)}")
print()

print("QLFS indicators:")
print(qlfs_2024_context.columns.tolist())

print()
print("QLFS Q1 2024 values:")
display(qlfs_2024_context)

assert len(qlfs_2024_context) == 1
assert qlfs_2024_context["Geography"].iloc[0] == "KwaZulu-Natal"

print()
print("QLFS Q1 2024 preparation: PASSED")
```

```python
#10. — Add QLFS Snapshot to Province Panel
# Purpose:
# Add the Q1 2024 labour-market snapshot to the province panel
# while keeping its timing explicitly separate from the historical years.

qlfs_snapshot = qlfs_2024_context.rename(
    columns={
        "Geography": "qlfs_geography",
        "Discouraged work-seekers": "QLFS_Q1_2024_DiscouragedWorkSeekers",
        "Employed": "QLFS_Q1_2024_Employed_thousand",
        "Employed/population ratio (absorption)": "QLFS_Q1_2024_AbsorptionRate",
        "Labour force": "QLFS_Q1_2024_LabourForce_thousand",
        "Labour force participation rate": "QLFS_Q1_2024_LabourForceParticipationRate",
        "Not economically active": "QLFS_Q1_2024_NotEconomicallyActive_thousand",
        "Other (not economically active)": "QLFS_Q1_2024_OtherNotEconomicallyActive_thousand",
        "Population 15–64 yrs": "QLFS_Q1_2024_Population15_64_thousand",
        "Unemployed": "QLFS_Q1_2024_Unemployed_thousand",
        "Unemployment rate": "QLFS_Q1_2024_UnemploymentRate"
    }
)

province_context_panel = province_context_panel.assign(
    QLFS_Q1_2024_Snapshot=True
)

province_context_panel = province_context_panel.merge(
    qlfs_snapshot,
    how="cross",
    validate="many_to_one"
)

print("QLFS Q1 2024 snapshot added.")
print(f"Rows: {len(province_context_panel):,}")
print(f"Columns: {len(province_context_panel.columns)}")
print()

print("QLFS snapshot timing:")
print("Q1 2024")

print()
print("QLFS snapshot values:")
display(
    province_context_panel[
        [
            "year",
            "QLFS_Q1_2024_Snapshot",
            "QLFS_Q1_2024_Employed_thousand",
            "QLFS_Q1_2024_Unemployed_thousand",
            "QLFS_Q1_2024_UnemploymentRate",
            "QLFS_Q1_2024_AbsorptionRate",
            "QLFS_Q1_2024_LabourForceParticipationRate"
        ]
    ]
)

assert len(province_context_panel) == 3
assert province_context_panel["QLFS_Q1_2024_Snapshot"].all()

print()
print("QLFS snapshot merge: PASSED")
```

```python
#10. — Store QLFS Q1 2024 Snapshot
# Purpose:
# Keep the QLFS labour-market snapshot separate from the historical
# province-year panel because it represents a specific Q1 2024 period.

qlfs_snapshot = qlfs_2024_context.rename(
    columns={
        "Geography": "QLFS_Geography",
        "Discouraged work-seekers": "QLFS_Q1_2024_DiscouragedWorkSeekers",
        "Employed": "QLFS_Q1_2024_Employed_thousand",
        "Employed/population ratio (absorption)": "QLFS_Q1_2024_AbsorptionRate",
        "Labour force": "QLFS_Q1_2024_LabourForce_thousand",
        "Labour force participation rate": "QLFS_Q1_2024_LabourForceParticipationRate",
        "Not economically active": "QLFS_Q1_2024_NotEconomicallyActive_thousand",
        "Other (not economically active)": "QLFS_Q1_2024_OtherNotEconomicallyActive_thousand",
        "Population 15–64 yrs": "QLFS_Q1_2024_Population15_64_thousand",
        "Unemployed": "QLFS_Q1_2024_Unemployed_thousand",
        "Unemployment rate": "QLFS_Q1_2024_UnemploymentRate"
    }
)

print("QLFS Q1 2024 snapshot prepared.")
print(f"Rows: {len(qlfs_snapshot):,}")
print(f"Columns: {len(qlfs_snapshot.columns):,}")
print()

print("QLFS snapshot:")
display(qlfs_snapshot)

assert len(qlfs_snapshot) == 1
assert qlfs_snapshot["QLFS_Geography"].iloc[0] == "KwaZulu-Natal"

print()
print("QLFS snapshot preparation: PASSED")
```

```python
#11. — Prepare Province Context Output
# Purpose:
# Combine the historical province-year panel with the clearly labelled
# QLFS Q1 2024 snapshot before final validation and saving.

province_context_output = province_context_panel.copy()

for column in qlfs_snapshot.columns:
    province_context_output[column] = qlfs_snapshot[column].iloc[0]

print("Province context output prepared.")
print(f"Rows: {len(province_context_output):,}")
print(f"Columns: {len(province_context_output.columns):,}")
print()

print("Historical years:")
print(sorted(province_context_output["year"].unique()))

print()
print("QLFS snapshot period: Q1 2024")

assert len(province_context_output) == 3
assert province_context_output["year"].tolist() == [2011, 2015, 2023]
assert province_context_output["QLFS_Geography"].eq("KwaZulu-Natal").all()

print()
print("Province context output preparation: PASSED")
```

```python
#12. — Validate Province Context Panel
# Purpose:
# Verify the province-year grain, KZN coverage and QLFS snapshot
# before saving the completed province context panel.

print("Province context validation:")
print(f"Rows: {len(province_context_output):,}")
print(f"Unique years: {province_context_output['year'].nunique():,}")
print(f"Unique province: {province_context_output['geography_name'].nunique():,}")

print()
print("Duplicate province-year records:")
print(
    province_context_output
    .duplicated(subset=["geography_name", "year"])
    .sum()
)

print()
print("Missing values by year:")
display(
    province_context_output[
        ["year", "geography_name"]
    ].assign(
        MissingValues=province_context_output.isna().sum(axis=1)
    )
)

assert len(province_context_output) == 3
assert province_context_output["geography_name"].eq("KwaZulu-Natal").all()
assert province_context_output["year"].nunique() == 3
assert province_context_output.duplicated(
    subset=["geography_name", "year"]
).sum() == 0

print()
print("Province context validation: PASSED")
```

The socioeconomic data covers 2011, 2015, and 2023, with values expected for each of these years. The youth distrust dataset contains observations only for 2023, so there are no equivalent 2011 or 2015 KZN observations. After merging the datasets into the province-year panel, we record the youth distrust values for 2011 and 2015 as `NaN`, while the 2023 values contain the available observations. We keep these values as `NaN` because missing data represents an unavailable observation, while `0% distrust` would represent an actual measured value. We apply the same approach to the QLFS snapshot, which is specifically from Q1 2024, so we treat those values as 2024 observations rather than assigning them to 2011, 2015, or 2023. During validation, the dataset shows five missing values for 2011 and 2015, so we will identify the specific five columns containing these missing values to confirm that they correspond only to the youth-distrust indicators before finalising Notebook 02.


```python
#13. — Save Province Context Panel
# Purpose:
# Save the validated KZN province context panel so it can be used
# as a processed input for the later merging stage.

province_context_output.to_csv(
    CONTEXT_OUTPUT,
    index=False
)

print("Province context panel saved successfully.")
print(f"Output file: {CONTEXT_OUTPUT}")
print(f"Rows saved: {len(province_context_output):,}")
print(f"Columns saved: {len(province_context_output.columns):,}")

assert CONTEXT_OUTPUT.exists(), "Output file was not cr
print()
print("Save validation: PASSED")
```
