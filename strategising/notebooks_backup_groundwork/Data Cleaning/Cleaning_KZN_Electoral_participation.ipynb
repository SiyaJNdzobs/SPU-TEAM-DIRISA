```python
from pathlib import Path
import pandas as pd

# Set the raw and cleaned data folders
BASE_DIR = Path(r"C:\Users\Student\Downloads\Big Data\SPU-TEAM-DIRISA")
RAW_DIR = BASE_DIR / "data" / "raw" / "KZN_Electoral_participation"
CLEAN_DIR = BASE_DIR / "data" / "interim" / "KZN_Electoral_participation Cleaned"

CLEAN_DIR.mkdir(parents=True, exist_ok=True)

# List the electoral participation files to inspect
files = [
    "P02111stQuarter2024.csv",
    "P03182024.csv",
    "Presentation-QLFS-Q1-2026.csv"
]

for filename in files:

    file_path = RAW_DIR / filename

    print("\n" + "=" * 80)
    print(filename)
    print("=" * 80)

    # Detect the file encoding before loading
    encodings = ["utf-8-sig", "utf-8", "utf-16", "latin1"]

    for encoding in encodings:
        try:
            df = pd.read_csv(
                file_path,
                encoding=encoding,
                low_memory=False
            )
            print(f"Encoding: {encoding}")
            print(f"Rows: {len(df):,}")
            print(f"Columns: {len(df.columns)}")
            print("Column names:")
            print(df.columns.tolist())
            print("\nData types:")
            print(df.dtypes)
            print("\nMissing values:")
            print(df.isna().sum())
            print("\nFirst 5 rows:")
            display(df.head())
            break

        except (UnicodeDecodeError, pd.errors.ParserError):
            continue
```

```python
# Inspect the raw file structure before choosing a cleaning method
# This prevents metadata and presentation rows from being treated as data

for filename in files:

    file_path = RAW_DIR / filename

    print("\n" + "=" * 90)
    print(f"{filename} — RAW STRUCTURE")
    print("=" * 90)

    with open(
        file_path,
        "r",
        encoding="utf-8-sig",
        errors="replace"
    ) as f:

        lines = [next(f, "").rstrip("\n\r") for _ in range(40)]

    for number, line in enumerate(lines, start=1):
        print(f"{number:03}: {line[:500]}")
```

```python
# Locate statistical tables and KZN references in the publication files
# This helps identify the relevant data sections before extraction

search_terms = [
    "KwaZulu-Natal",
    "KZN",
    "Table",
    "Province",
    "Employment",
    "Unemployment",
    "labour force",
    "voting",
    "participation"
]

for filename in files:

    file_path = RAW_DIR / filename

    print("\n" + "=" * 90)
    print(f"{filename} — RELEVANT SECTIONS")
    print("=" * 90)

    with open(
        file_path,
        "r",
        encoding="utf-8-sig",
        errors="replace"
    ) as f:

        lines = f.readlines()

    matches = []

    for number, line in enumerate(lines, start=1):
        clean_line = line.strip()

        if any(
            term.lower() in clean_line.lower()
            for term in search_terms
        ):
            matches.append((number, clean_line))

    print(f"Matching lines found: {len(matches):,}")

    for number, line in matches[:100]:
        print(f"{number:04}: {line[:500]}")
```

```python
# Locate the province-level labour-force table in P0211
# We extract the surrounding rows before deciding the final table structure

filename = "P02111stQuarter2024.csv"
file_path = RAW_DIR / filename

with open(
    file_path,
    "r",
    encoding="utf-8-sig",
    errors="replace"
) as f:
    lines = f.readlines()

# Find the province-level table references
matches = [
    (i + 1, line.strip())
    for i, line in enumerate(lines)
    if "Table 2.3" in line or "KwaZulu-Natal" in line
]

print(f"Relevant lines found: {len(matches):,}")

for number, line in matches[:80]:
    print(f"{number:04}: {line[:500]}")
```

```python
# Extract the KZN section of Table 2.3
# We inspect the headers and KZN rows before building the cleaned dataset

start = 1230
end = 1730

print("=" * 100)
print("P0211 Q1 2024 — TABLE 2.3: LABOUR FORCE CHARACTERISTICS")
print("=" * 100)

for number in range(start, end + 1):

    line = lines[number - 1].strip()

    if line:
        print(f"{number:04}: {line}")
```

```python
# Extract the KZN rows from Table 2.3
# Keep the province, non-metro and eThekwini areas for cleaning

kzn_rows = []

for number in range(1505, 1561):

    line = lines[number - 1].strip()

    if line:
        kzn_rows.append(line)

print("=" * 100)
print("KZN TABLE 2.3 ROWS")
print("=" * 100)

for row in kzn_rows:
    print(row)
```

```python
import re

# Convert the extracted KZN Table 2.3 rows into a structured dataset
# Keep the five quarterly values and the reported change measures

records = []

metric_names = [
    "Population 15–64 yrs",
    "Labour force",
    "Employed",
    "Unemployed",
    "Not economically active",
    "Discouraged work-seekers",
    "Other (not economically active)",
    "Unemployment rate",
    "Employed/population ratio (absorption)",
    "Labour force participation rate"
]

geographies = [
    "KwaZulu-Natal",
    "KwaZulu-Natal – Non-metro",
    "KwaZulu-Natal – eThekwini"
]

current_geo = None

for row in kzn_rows:

    row = row.strip('"')

    if row in geographies:
        current_geo = row
        continue

    for metric in metric_names:

        if row.startswith(metric):

            values_text = row[len(metric):].strip()

            values = re.findall(
                r"-?\d+(?:[ ]\d{3})*(?:,\d+)?",
                values_text
            )

            values = [
                float(value.replace(" ", "").replace(",", "."))
                for value in values
            ]

            records.append({
                "Geography": current_geo,
                "Metric": metric,
                "Q1_2023": values[0],
                "Q2_2023": values[1],
                "Q3_2023": values[2],
                "Q4_2023": values[3],
                "Q1_2024": values[4],
                "Qtr_to_qtr_change": values[5],
                "Year_on_year_change": values[6],
                "Qtr_to_qtr_percent": values[7] if len(values) > 7 else None,
                "Year_on_year_percent": values[8] if len(values) > 8 else None
            })

            break

kzn_clean = pd.DataFrame(records)

print("=" * 100)
print("KZN TABLE 2.3 STRUCTURED DATA")
print("=" * 100)

print(f"Rows: {len(kzn_clean)}")
print(f"Columns: {len(kzn_clean.columns)}")

display(kzn_clean)
```

```python
# Correct the shortened rate metric names caused by overlapping metric labels
# Keep the original values unchanged

kzn_clean.loc[
    kzn_clean["Metric"] == "Employed",
    "Metric"
] = "Employed/population ratio (absorption)"

kzn_clean.loc[
    kzn_clean["Metric"] == "Labour force",
    "Metric"
] = "Labour force participation rate"

print("=" * 100)
print("CORRECTED METRICS")
print("=" * 100)

print(kzn_clean["Metric"].value_counts())
```

```python
# Restore the correct metric names using their fixed Table 2.3 order
# Each geography contains seven count measures followed by three rate measures

metric_order = [
    "Population 15–64 yrs",
    "Labour force",
    "Employed",
    "Unemployed",
    "Not economically active",
    "Discouraged work-seekers",
    "Other (not economically active)",
    "Unemployment rate",
    "Employed/population ratio (absorption)",
    "Labour force participation rate"
]

kzn_clean["Metric"] = (
    kzn_clean.groupby("Geography").cumcount()
)

kzn_clean["Metric"] = kzn_clean["Metric"].map(
    dict(enumerate(metric_order))
)

print("=" * 100)
print("FINAL METRIC CHECK")
print("=" * 100)

print(kzn_clean["Metric"].value_counts())
```

```python
# Validate the structured KZN labour-force dataset
# Check duplicates, missing values, geography coverage and expected row count

print("=" * 100)
print("KZN TABLE 2.3 VALIDATION")
print("=" * 100)

print(f"Expected rows: 30")
print(f"Actual rows:   {len(kzn_clean)}")
print(f"Row count correct: {len(kzn_clean) == 30}")

print("\nExpected geographies:")
print(kzn_clean["Geography"].unique())

print("\nMissing values:")
print(kzn_clean.isna().sum())

print("\nDuplicate rows:")
print(kzn_clean.duplicated().sum())

print("\nMetric counts:")
print(kzn_clean["Metric"].value_counts())

print("\nGeography counts:")
print(kzn_clean["Geography"].value_counts())
```

```python
# Finalise the cleaned KZN Table 2.3 dataset
# Save the cleaned file without modifying the original raw publication

output_file = CLEAN_DIR / "P02111stQuarter2024_clean.csv"

kzn_clean.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("=" * 100)
print("CLEANED DATASET SAVED")
print("=" * 100)

print(f"File: {output_file}")
print(f"Rows: {len(kzn_clean):,}")
print(f"Columns: {len(kzn_clean.columns)}")
print(f"File exists: {output_file.exists()}")
```
