# Data Cleaning: KZN Socioeconomic and Political Distrust 2011–2023

This notebook cleans and validates the KwaZulu-Natal socioeconomic and political distrust datasets.

The raw files are kept unchanged in `data/raw/`. The cleaned versions are saved separately in `data/interim/`.

The cleaning process checks for:
- File structure
- Missing values
- Blank values
- Duplicate records
- Inconsistent formats
- Data types
- Year consistency
- Province and location values
- Invalid or unexpected values
- Preservation of records after cleaning

```python
import pandas as pd
from pathlib import Path
```

```python
# Notebook location:
# SPU-TEAM-DIRISA/notebooks/Data Collection/

project_root = Path("../..")

raw_folder = (
    project_root
    / "data"
    / "raw"
    / "KZN-Socioeconomic-Political-Distrust-2011-2023"
)

interim_folder = (
    project_root
    / "data"
    / "interim"
    / "KZN-Socioeconomic-Political-Distrust-2011-2023"
)

interim_folder.mkdir(parents=True, exist_ok=True)

print("Raw folder:", raw_folder)
print("Raw folder exists:", raw_folder.exists())
print("Interim folder:", interim_folder)
```

```python
raw_files = sorted(
    [file for file in raw_folder.iterdir() if file.is_file()]
)

print("Number of files:", len(raw_files))

for file in raw_files:
    print(file.name)
```

```python
# Load the two datasets

dataframes = {}

for file in raw_files:

    if file.suffix.lower() == ".csv":
        df = pd.read_csv(file)
    
    elif file.suffix.lower() in [".xlsx", ".xls"]:
        df = pd.read_excel(file)
    
    else:
        print("Unsupported file type:", file.name)
        continue

    dataframes[file.name] = df

    print("\nFile:", file.name)
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
```

```python
for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    print("\nColumns:")
    for number, column in enumerate(df.columns, start=1):
        print(f"{number}. {column}")

    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# Check the structure of each dataset

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    print("Number of rows:", len(df))
    print("Number of columns:", len(df.columns))
    print("Column names and data types:")

    display(
        pd.DataFrame({
            "column": df.columns,
            "data_type": df.dtypes.astype(str).values
        })
    )
```

```python
# Check for missing values

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    missing = df.isnull().sum()

    display(missing.to_frame("missing_values"))

    print("Total missing values:", missing.sum())
```

```python
# Check for blank or empty values

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    blank_counts = {}

    for column in df.columns:

        if df[column].dtype == "object":
            blank_counts[column] = (
                df[column]
                .astype(str)
                .str.strip()
                .eq("")
                .sum()
            )
        else:
            blank_counts[column] = 0

    blank_counts = pd.Series(blank_counts)

    display(blank_counts.to_frame("blank_values"))

    print("Total blank values:", blank_counts.sum())
```

```python
# Check for duplicate complete records

for file_name, df in dataframes.items():

    duplicate_count = df.duplicated().sum()

    print(
        file_name,
        "| Duplicate complete records:",
        duplicate_count
    )
```

```python
# Check for unnecessary spaces in text values

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    for column in df.select_dtypes(include="object").columns:

        spaces = (
            df[column]
            .astype(str)
            .ne(df[column].astype(str).str.strip())
            .sum()
        )

        print(
            f"{column} | Values with leading/trailing spaces: {spaces}"
        )
```

```python
# Check unique values in text columns

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    for column in df.select_dtypes(include="str").columns:

        unique_values = df[column].dropna().unique()

        print(f"\n{column}")
        print("Number of unique values:", len(unique_values))

        for value in sorted(unique_values.astype(str)):
            print(" -", value)
```

```python
# Inspect numerical columns

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    numeric_columns = df.select_dtypes(include="number").columns

    if len(numeric_columns) == 0:
        print("No numeric columns detected.")
    else:
        display(df[numeric_columns].describe().T)
```

```python
# Check columns that appear to contain year information

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    year_columns = [
        column
        for column in df.columns
        if "year" in column.lower()
    ]

    if year_columns:
        for column in year_columns:
            print(f"\n{column}")
            print(
                df[column]
                .dropna()
                .unique()
            )
    else:
        print("No column containing 'year' found.")
```

```python
# Check province and location-related columns

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    location_columns = [
        column
        for column in df.columns
        if any(
            word in column.lower()
            for word in [
                "province",
                "municipality",
                "district",
                "location",
                "region",
                "area"
            ]
        )
    ]

    if location_columns:

        for column in location_columns:

            print(f"\n{column}")
            print(
                df[column]
                .dropna()
                .astype(str)
                .unique()
            )

    else:
        print("No obvious location-related columns found.")
```

```python
# Check geography and location values directly

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    location_columns = [
        column
        for column in df.columns
        if any(
            word in column.lower()
            for word in [
                "geography",
                "province",
                "municipality",
                "district",
                "location",
                "region",
                "area"
            ]
        )
    ]

    if location_columns:

        for column in location_columns:

            print(f"\n{column}")
            print("Number of unique values:", df[column].nunique())

            values = df[column].dropna().astype(str).unique()

            for value in sorted(values):
                print(" -", value)

    else:
        print("No geography or location-related columns found.")
```

```python
# Look for common indicators of possible junk or formatting problems

possible_problem_values = [
    "",
    " ",
    "N/A",
    "NA",
    "n/a",
    "na",
    "null",
    "NULL",
    "-",
    "--"
]

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    for column in df.select_dtypes(include="object").columns:

        found = (
            df[column]
            .astype(str)
            .str.strip()
            .isin(possible_problem_values)
            .sum()
        )

        if found > 0:
            print(
                f"{column} | Possible placeholder/junk values: {found}"
            )
```

```python
# Check for completely empty rows

for file_name, df in dataframes.items():

    empty_rows = df.isnull().all(axis=1).sum()

    print(
        file_name,
        "| Completely empty rows:",
        empty_rows
    )
```

```python
# Check for completely empty columns

for file_name, df in dataframes.items():

    empty_columns = df.isnull().all(axis=0)

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    print(
        "Completely empty columns:",
        empty_columns.sum()
    )

    if empty_columns.any():
        print(
            list(df.columns[empty_columns])
        )
```

```python
# Check text columns for values that appear to be numbers

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    for column in df.select_dtypes(include="object").columns:

        cleaned_values = (
            df[column]
            .dropna()
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
        )

        if len(cleaned_values) > 0:

            numeric_conversion = pd.to_numeric(
                cleaned_values,
                errors="coerce"
            )

            numeric_like = numeric_conversion.notna().sum()

            if numeric_like > 0:
                print(
                    f"{column} | Numeric-looking text values: "
                    f"{numeric_like} of {len(cleaned_values)}"
                )
```

```python
# Check text columns for values that appear to be numbers

for file_name, df in dataframes.items():

    print("\n" + "=" * 80)
    print("FILE:", file_name)
    print("=" * 80)

    for column in df.select_dtypes(include="object").columns:

        cleaned_values = (
            df[column]
            .dropna()
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
        )

        if len(cleaned_values) > 0:

            numeric_conversion = pd.to_numeric(
                cleaned_values,
                errors="coerce"
            )

            numeric_like = numeric_conversion.notna().sum()

            if numeric_like > 0:
                print(
                    f"{column} | Numeric-looking text values: "
                    f"{numeric_like} of {len(cleaned_values)}"
                )
```

```python
# Create copies for cleaning

cleaned_dataframes = {}

for file_name, df in dataframes.items():

    cleaned_df = df.copy()

    # Remove leading/trailing spaces from text columns
    for column in cleaned_df.select_dtypes(include="object").columns:
        cleaned_df[column] = cleaned_df[column].str.strip()

    cleaned_dataframes[file_name] = cleaned_df

print(
    "Cleaned copies created for:",
    len(cleaned_dataframes),
    "files"
)
```

```python
# Save cleaned files to the interim folder

for file_name, cleaned_df in cleaned_dataframes.items():

    output_file = interim_folder / file_name

    if output_file.suffix.lower() == ".csv":
        cleaned_df.to_csv(
            output_file,
            index=False,
            encoding="utf-8"
        )

    elif output_file.suffix.lower() in [".xlsx", ".xls"]:
        cleaned_df.to_excel(
            output_file,
            index=False
        )

    print("Saved:", output_file)
```

```python
# Reload the cleaned files

validated_dataframes = {}

for file_name, cleaned_df in cleaned_dataframes.items():

    cleaned_file = interim_folder / file_name

    if cleaned_file.suffix.lower() == ".csv":
        validated_df = pd.read_csv(cleaned_file)

    elif cleaned_file.suffix.lower() in [".xlsx", ".xls"]:
        validated_df = pd.read_excel(cleaned_file)

    validated_dataframes[file_name] = validated_df

    print(
        file_name,
        "| Rows:", len(validated_df),
        "| Columns:", len(validated_df.columns)
    )
```

```python
# Final validation

print("=" * 60)
print("FINAL CLEANING VALIDATION")
print("=" * 60)

for file_name in dataframes:

    original = dataframes[file_name]
    cleaned = validated_dataframes[file_name]

    print("\nFILE:", file_name)

    print("Original rows:", len(original))
    print("Cleaned rows:", len(cleaned))

    print(
        "Rows preserved:",
        len(original) == len(cleaned)
    )

    print(
        "Original columns:",
        len(original.columns)
    )

    print(
        "Cleaned columns:",
        len(cleaned.columns)
    )

    print(
        "Columns preserved:",
        list(original.columns) == list(cleaned.columns)
    )

    print(
        "Missing values after cleaning:",
        cleaned.isnull().sum().sum()
    )

    print(
        "Duplicate records after cleaning:",
        cleaned.duplicated().sum()
    )
```

```python
# Confirm that every cleaned file exists

print("Expected files:", len(raw_files))

saved_files = sorted(
    [file for file in interim_folder.iterdir() if file.is_file()]
)

print("Saved files:", len(saved_files))

for file in saved_files:
    print(file.name)
```
