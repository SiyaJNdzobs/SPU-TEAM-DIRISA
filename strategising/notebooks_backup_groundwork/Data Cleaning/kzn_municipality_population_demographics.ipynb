```python
import pandas as pd
from pathlib import Path

# Load the raw file
raw_file = Path(
    r"C:\Users\Admin\PycharmProjects\SPU-TEAM-DIRISA\notebooks\data\raw\KZN population demographics\kzn_municipality_population_demographics.csv"
)

df = pd.read_csv(raw_file)

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Remove unnecessary spaces from municipality names
df["municipality"] = df["municipality"].str.strip()

print("===== CLEANED COLUMNS =====")
print(df.columns.tolist())

print("\n===== MUNICIPALITIES =====")
print(df["municipality"].tolist())
```

```python
print("===== MUNICIPALITY DUPLICATES =====")

duplicate_municipalities = df[
    df["municipality"].duplicated(keep=False)
]

print(duplicate_municipalities)

print("\nNumber of duplicate municipality names:",
      df["municipality"].duplicated().sum())
```

```python
from pathlib import Path
import pandas as pd

# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Clean municipality names
df["municipality"] = df["municipality"].str.strip()

# Numeric columns
numeric_columns = [
    "household",
    "homeless",
    "transient",
    "institution",
    "urban_area",
    "tribal_or_traditional_area",
    "farm_area",
    "male",
    "population"
]

# Make sure numeric columns are numeric
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Create interim folder
interim_folder = Path(
    r"C:\Users\Admin\PycharmProjects\SPU-TEAM-DIRISA\notebooks\data\interim\KZN population demographics"
)

interim_folder.mkdir(parents=True, exist_ok=True)

# Save cleaned file
interim_file = interim_folder / "kzn_municipality_population_demographics.csv"

df.to_csv(interim_file, index=False)

print("Cleaned file saved successfully!")
print(interim_file)
print("\nShape:", df.shape)
print("\nMissing values after cleaning:")
print(df.isna().sum())
```
